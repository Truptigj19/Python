"""
## PANDAS - JOIN / MERGE AND CONCATENATION

 merge()
-------
Used to combine DataFrames using a common column/key.

concat()
--------
Used to combine multiple Pandas objects along rows or columns.

Important merge types:
- inner
- left
- right
- outer
"""

```python
import pandas as pd
```

```python
customers = {
    "Customer_ID": [101, 102, 103, 104],
    "Name": ["Amit", "Sneha", "Rahul", "Priya"],
    "City": ["Pune", "Mumbai", "Delhi", "Nashik"]
}
```

```python
orders = {
    "Customer_ID": [101, 102, 103, 105],
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Amount": [65000, 42000, 30000, 18000]
}
```

```python
customers_df = pd.DataFrame(customers)
orders_df = pd.DataFrame(orders)
```

```python
print("\ncustomers_Table")
customers_df
```

```python
print("\norders_df")
orders_df
```

PART 1: MERGE

```python
# Inner Merge
inner_merge = pd.merge(
    customers_df,
    orders_df,
    on="Customer_ID",
    how="inner"
)

print("INNER MERGE:")
print(inner_merge)
```

```python
# Left Merge
left_merge = pd.merge(
    customers_df,
    orders_df,
    on="Customer_ID",
    how="left"
)

print("\nLEFT MERGE:")
print(left_merge)
```

```python
# Outer Merge
outer_merge = pd.merge(
    customers_df,
    orders_df,
    on="Customer_ID",
    how="outer"
)

print("\nOUTER MERGE:")
print(outer_merge)
```
