```python
import pandas as pd
```

Creating the dataset

```python
orders = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Customer": ["Amit", "Sneha", "Rahul", "Priya",
                 "Amit", "Neha", "Sneha", "Om"],
    "Product": ["Laptop", "Phone", "Mouse", "Keyboard",
                "Phone", "Laptop", "Mouse", "Monitor"],
    "Category": ["Electronics", "Electronics", "Accessories",
                 "Accessories", "Electronics", "Electronics",
                 "Accessories", "Electronics"],
    "Amount": [65000, 42000, 800, 1500, 45000, 72000, 900, 18000],
    "Quantity": [1, 1, 2, 1, 1, 1, 3, 1]
}
```

```python
df = pd.DataFrame(orders)
```

```python
df
```

```python
#head()
df.head(2)
```

```python
#tail()
df.tail(3)
```

DataFrame Attributes

```python
print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nIndex:")
print(df.index)

print("\nData Types:")
print(df.dtypes)

print("\nTotal Elements:")
print(df.size)

print("\nNumber of Dimensions:")
print(df.ndim)
```

Statistical Summary :

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
#Count of Non-Null Values
print("\nNon-Null Count:")
print(df.count())


# --------------------------------------------------
# Number of Unique Customers
# --------------------------------------------------

print("\nNumber of Unique Customers:")
print(df["Customer"].nunique())


# --------------------------------------------------
# DataFrame Information
# --------------------------------------------------

print("\nDataFrame Information:")
df.info()
```
